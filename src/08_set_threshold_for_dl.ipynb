07_train_vitaldb_model.py에서 학습된 딥러닝 모델들의 Specificity 80/90/95% threshold cutoff 찾기

```python
from pathlib import Path
import numpy as np
import numpy.typing as npt
from keras.models import load_model
from src.module.utils import load_config
```

```python
def scale_input(input) -> npt.NDArray:
    return input.astype(np.float32) / 200
```

```python
config = load_config("../config/conf.yaml")
src_path = Path(config["data_path"]).expanduser() / "vitaldb" / "06.train_val_test_new"
model_path = Path(config["model_path"]).expanduser() / "revision"
strategy = "hypophetversion2"
```

```python
val_X = np.load(src_path / f"vitaldb_{strategy}_val_X.npy")
val_y = np.load(src_path / f"vitaldb_{strategy}_val_y.npy")
print(val_X.shape, val_y.shape)
```

```python
val_X_s = scale_input(val_X)
```

```python
lodaed_model = load_model(model_path / "inceptiontime.model.keras")
```

```python
prob_y = lodaed_model.predict(val_X_s)
```

```python
neg = prob_y[np.where(val_y == 0)]
for spe in [0.80, 0.90, 0.95]:
    cutoff = np.quantile(neg, spe)
    print(f"{spe}, {cutoff:.2f}")
```
